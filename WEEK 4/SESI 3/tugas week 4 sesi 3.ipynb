```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()
```

Penjelasan:
Pada tahap pertama, beberapa library yang dibutuhkan diimpor terlebih dahulu. Library NumPy digunakan untuk mengolah data dalam bentuk array, sedangkan Pandas digunakan untuk membuat dan mengolah data dalam bentuk DataFrame. Selain itu, `load_iris` dari Scikit-learn digunakan untuk mengambil dataset Iris yang sudah tersedia di dalam library Scikit-learn.
Dataset Iris nantinya digunakan sebagai data untuk melakukan klasifikasi menggunakan algoritma Naïve Bayes.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]

print(X)
```

Penjelasan: Mengubah Dataset Menjadi DataFrame dan Menentukan Fitur serta Target
Pada tahap ini data dari dataset Iris diubah menjadi DataFrame menggunakan Pandas agar lebih mudah untuk dilihat dan diolah.
`iris.data` berisi nilai dari empat fitur pada bunga Iris, sedangkan `iris.feature_names` digunakan sebagai nama kolomnya. Setelah itu dibuat kolom baru bernama `Species` yang berisi nama species dari setiap data bunga Iris.

Selanjutnya data dipisahkan menjadi dua bagian. Variabel `X` digunakan sebagai fitur atau data masukan yang terdiri dari empat karakteristik bunga, yaitu sepal length, sepal width, petal length, dan petal width. Sedangkan variabel `y` digunakan sebagai target atau label yang berisi species bunga.
Kolom `Species` tidak dimasukkan ke dalam `X` karena kolom tersebut merupakan hasil atau kelas yang nantinya akan diprediksi oleh model.


```python
y
```

Penjelasan:

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

Penjelasan:Mengubah Label Species Menjadi Angka

Pada tahap ini label species yang sebelumnya masih berupa teks diubah menjadi bentuk angka menggunakan `LabelEncoder`.
Sebelumnya, nilai pada `y` masih berupa nama species, yaitu `setosa`, `versicolor`, dan `virginica`. Agar dapat digunakan dalam proses klasifikasi, label tersebut diubah menjadi angka.
`LabelEncoder()` akan memberikan kode angka untuk setiap kelas. Setelah proses `fit_transform()` dilakukan, label species menjadi nilai numerik, yaitu 0, 1, dan 2. Nilai tersebut tetap mewakili species yang sama, hanya bentuk datanya saja yang diubah menjadi angka.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

Penjelasan: Membagi Data Training dan Data Testing

Pada tahap ini dataset dibagi menjadi data training dan data testing menggunakan `train_test_split()`.
Data training digunakan untuk melatih model Naïve Bayes agar model dapat mempelajari pola dari data yang tersedia. Sedangkan data testing digunakan untuk menguji hasil prediksi model menggunakan data yang tidak digunakan saat proses pelatihan.
Parameter `test_size=0.2` berarti 20% data digunakan sebagai data testing dan 80% sisanya digunakan sebagai data training. `random_state=0` digunakan agar pembagian data tetap sama ketika kode dijalankan kembali.
Pembagian dilakukan untuk `X` dan `y` sehingga data fitur dan labelnya tetap digunakan sesuai dengan kebutuhan model.

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

Penjelasan: Membuat dan Melatih Model Gaussian Naïve Bayes

Pada tahap ini digunakan algoritma Gaussian Naïve Bayes untuk melakukan klasifikasi dataset Iris.

Gaussian Naïve Bayes digunakan karena fitur pada dataset Iris berupa data numerik kontinu. Model dibuat menggunakan `GaussianNB()` kemudian dilatih menggunakan data training yang sudah dibuat sebelumnya.

`model.fit(X_train, y_train)` digunakan untuk melatih model berdasarkan data fitur dan label pada data training. Setelah model selesai dilatih, `predict(X_test)` digunakan untuk melakukan prediksi terhadap data testing.


```python
result = NB_model.predict(X_test)
result
```

Penjelasan:


```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

Penjelasan: Menghitung Accuracy Model

Setelah model melakukan prediksi terhadap data testing, langkah berikutnya adalah menghitung tingkat akurasi dari hasil prediksi tersebut.
Pada bagian ini digunakan `accuracy_score()` untuk membandingkan label sebenarnya pada `y_test` dengan hasil prediksi yang terdapat pada `result`.

Hasil akurasi kemudian dikalikan dengan 100 agar ditampilkan dalam bentuk persentase. Dari hasil pengujian pada modul, diperoleh accuracy sebesar 96.66666666666667%, sehingga sebagian besar data testing berhasil diklasifikasikan dengan benar oleh model.

```python
#prediksi inputan data baru
#input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Penjelasan:  Melakukan Prediksi Data Baru

Setelah model selesai dilatih dan diuji, model kemudian digunakan untuk melakukan prediksi terhadap satu data baru.
Pada contoh di modul, data baru yang digunakan adalah `[5.1, 3.5, 1.4, 0.2]`. Data tersebut berisi empat nilai yang sesuai dengan empat fitur pada dataset Iris, yaitu sepal length, sepal width, petal length, dan petal width.
Data baru kemudian dimasukkan ke dalam `NB_model.predict()` untuk mengetahui kelas atau species yang diprediksi oleh model.

Karena hasil prediksi dari model masih berupa angka, digunakan `lab.inverse_transform()` untuk mengubah kembali angka tersebut menjadi nama species agar hasilnya lebih mudah dipahami.

```python
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Penjelasan:  Prediksi Data Baru Menggunakan DataFrame

Pada tahap ini data input yang sebelumnya berbentuk array diubah menjadi DataFrame. Empat nilai input diberikan nama kolom yang sama dengan fitur pada dataset Iris.
Setelah data dibuat dalam bentuk DataFrame, data tersebut dimasukkan ke dalam model `NB_model` untuk mendapatkan hasil prediksi.

Hasil prediksi yang masih berupa angka kemudian dikembalikan ke nama species menggunakan `inverse_transform()`. Dengan cara ini, hasil prediksi dapat langsung ditampilkan dalam bentuk nama species.
